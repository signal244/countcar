# BEV 변환 전용

이 노트북은 원본 영상을 BEV mp4로 변환만 합니다.
완성된 `_bev.mp4`는 이후 별도 탐지/추적 코랩에서 입력으로 사용하면 됩니다.

```python
from google.colab import drive
from pathlib import Path
import os
import subprocess
import sys

drive.mount('/content/drive', force_remount=True)

# 본인 값으로 수정
PROJECT_NAME = '창원공사중(분석용)'
JUNCTION_NAME = '상남사거리'
SESSION_NAME = '점심첨두'

# 본인 Drive 실제 경로로 수정
ROOT = Path('/content/drive/MyDrive/vm/count_car_ver5.5')
assert ROOT.exists(), f'프로젝트 경로 없음: {ROOT}'

script_path = ROOT / 'colab' / 'notebook_snippet_bev_only.py'
assert script_path.exists(), f'스크립트 경로 없음: {script_path}'

env = dict(os.environ)
prev = env.get('PYTHONPATH', '')
env['PYTHONPATH'] = f"{ROOT}:{prev}" if prev else str(ROOT)
env['COUNTCAR_PROJECT_NAME'] = PROJECT_NAME
env['COUNTCAR_JUNCTION_NAME'] = JUNCTION_NAME
env['COUNTCAR_SESSION_NAME'] = SESSION_NAME

proc = subprocess.Popen(
    [sys.executable, str(script_path)],
    cwd=str(ROOT),
    env=env,
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    bufsize=1,
)

assert proc.stdout is not None
for line in proc.stdout:
    print(line, end='')

ret = proc.wait()
if ret != 0:
    raise RuntimeError(f'notebook_snippet_bev_only.py failed with exit code {ret}')
```
